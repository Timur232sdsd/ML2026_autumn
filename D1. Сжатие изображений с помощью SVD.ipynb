## 2. Теоретическая справка (кратко)
Любую матрицу $A$ размером $m \times n$ можно разложить на три матрицы:
$$
A = U \cdot \Sigma \cdot V^T
$$
Где:
- $U$ — ортогональная матрица размера $m \times m$ (левые сингулярные векторы).
- $\Sigma$ — диагональная матрица размера $m \times n$, на диагонали которой стоят сингулярные числа $\sigma_1 \ge \sigma_2 \ge ... \ge \sigma_k$.
- $V^T$ — ортогональная матрица размера $n \times n$ (правые сингулярные векторы).

**Идея сжатия:** Если оставить только первые $r$ сингулярных чисел (и соответствующие столбцы U и V), мы получим матрицу $A_r$, которая является наилучшим приближением $A$ среди всех матриц ранга $r$ (в смысле нормы Фробениуса).

### Выполнение задания

````python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from sklearn.metrics import mean_squared_error

# Cчитывание данных
img = Image.open('images.jpg')
# color transforms
img_rgb = img.convert('RGB')
# conversion to an array
img_array = np.array(img_rgb, dtype=np.float32)
original_shape = img_array.shape
# conversion to an list
original_flat = img_array.astype(np.float64).flatten()
````

### Задание 3 Часть A
Считываем все необходимые данные

````python
# Изначальный вес
h, w, c = img_array.shape
original_size = c * h * w
# SVD разложение для каждого канала (RGB)
def svd(img_array, r):
    # height, width, color
    h, w, c = img_array.shape
    reconstructed = np.zeros_like(img_array, dtype=np.float32)
    for i in range(c):
        U, s, Vt = np.linalg.svd(img_array[:, :, i], full_matrices=False)
        reconstructed[:, :, i] = (U[:, :r] * s[:r]) @ Vt[:r, :]
    return np.clip(reconstructed, 0, 255)

# Количество сингулярных значений на главной диагонали 
list_of_ranks = [1, 5, 10, 30, 100, min(h, w)]
# Массивы изображений 
approximations = []
# Размер после SVD разложения 
compressed_sizes = []

for r in list_of_ranks:
    approx = svd(img_array, r)
    approximations.append(approx)
    # Считывание нового размера в байтах 
    compressed_size = r * (h + 1 + w) * 3 * 4
    compressed_sizes.append(original_size/compressed_size)
````

### Задание 3 часть Б, Г

Была написана функция SVD разложения матрицы для массива изображения, создан массив с рангами сингулярного разложения *list_of_ranks*. В массив *approximations* были добавлены SVD разложения матрицы при различных рангах разложения. Для каждого ранга был вычислен размер соответствующей матрицы в байтах. 

````python
# создание графика
fig, axs = plt.subplots(2, 3, figsize=(16, 10))
axs = axs.flatten()
# Визуализация 
for i, r in enumerate(list_of_ranks):
    axs[i].imshow(approximations[i].astype(np.uint8))
    axs[i].set_title(f"Ранг = {r}, сжатие в {compressed_sizes[i]:.1f} раза")

plt.show()
````

Заметим, что при ранге *r = 100* изображение визуально не отличается от максимально возможного.

````python
# ---------- БЛОК АВТОПРОВЕРКИ (НЕ РЕДАКТИРОВАТЬ) ----------
for i, r in enumerate(list_of_ranks):
    assert approximations[i].shape == original_shape, \
        f"Ошибка: размерность для ранга {r} не совпадает с исходной"

for i, r in enumerate(list_of_ranks):
    arr = approximations[i]
    if arr.dtype != np.uint8:
        arr = np.clip(arr, 0, 255)
    assert arr.max() <= 255.1, f"Значения превышают 255 для ранга {r}"
    assert arr.min() >= -0.1, f"Значения меньше 0 для ранга {r}"

mse_list = []
for i in range(len(approximations)):
    mse = mean_squared_error(original_flat, approximations[i].flatten())
    mse_list.append(mse)

for i in range(1, len(mse_list)):
    assert mse_list[i] <= mse_list[i-1] + 1e-6, \
        f"MSE не уменьшается между рангами {list_of_ranks[i-1]} и {list_of_ranks[i]}"

print("✅ Все автоматические проверки пройдены. Задание выполнено корректно!")
````
